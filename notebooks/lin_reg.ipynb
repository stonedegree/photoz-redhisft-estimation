```python
import numpy as np
import pandas as pd
import plotly.express as px
import metrics
import pickle

from matplotlib import pyplot as plt
from astroML.datasets import fetch_sdss_specgals
from sklearn.linear_model import LinearRegression
```

```python
with open('data/dataset_split_v1.pkl', 'rb') as file:
    loaded_data_v1 = pickle.load(file)
    
with open('data/dataset_split_v2.pkl', 'rb') as file:
    loaded_data_v2 = pickle.load(file)
    
with open('data/dataset_split_v3.pkl', 'rb') as file:
    loaded_data_v3 = pickle.load(file)
    
with open('data/dataset_split_v4.pkl', 'rb') as file:
    loaded_data_v4 = pickle.load(file)
```

```python
X_train, X_val, X_test, y_train, y_val, y_test = loaded_data_v1['retval']

lin_model_v1 = LinearRegression().fit(X_train, y_train)
z_pred_val_v1 = lin_model_v1.predict(X_val)

print("Outlier Fraction: ", metrics.outlier_fraction(z_pred_val_v1, y_val))
print("Bias:             ", metrics.bias(z_pred_val_v1, y_val))
print("NMAD:             ", metrics.nmad(z_pred_val_v1, y_val))
```
