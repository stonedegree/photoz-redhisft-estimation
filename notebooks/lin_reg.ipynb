```python
import numpy as np
import pandas as pd
import plotly.express as px
import metrics
import pickle

from matplotlib import pyplot as plt
from astroML.datasets import fetch_sdss_specgals
from sklearn.linear_model import LinearRegression
```

```python
with open('data/dataset_split_v1.pkl', 'rb') as file:
    loaded_data_v1 = pickle.load(file)
    
with open('data/dataset_split_v2.pkl', 'rb') as file:
    loaded_data_v2 = pickle.load(file)
    
with open('data/dataset_split_v3.pkl', 'rb') as file:
    loaded_data_v3 = pickle.load(file)
    
with open('data/dataset_split_v4.pkl', 'rb') as file:
    loaded_data_v4 = pickle.load(file)
```

```python
# Dictionary to store trained models
trained_models = {}

# Iterate through versions
for v in ['v1', 'v2', 'v3', 'v4']:
    print(f"Results for Dataset {v}: ")
    
    # Load the corresponding dataset
    with open(f'data/dataset_split_{v}.pkl', 'rb') as file:
        loaded_data = pickle.load(file)
        
    # Unpack the data
    X_train, X_val, X_test, y_train, y_val, y_test = loaded_data['retval']
    
    # Initialize and fit the Linear Regression model
    lin_model = LinearRegression().fit(X_train, y_train)
    
    # Save the model to dictionary
    trained_models[v] = lin_model
    
    # Make predictions on the validation set
    z_pred_val = lin_model.predict(X_val)
    
    # Calculate and print the metrics
    print("Outlier Fraction: ", metrics.outlier_fraction(z_pred_val, y_val))
    print("Bias:             ", metrics.bias(z_pred_val, y_val))
    print("NMAD:             ", metrics.nmad(z_pred_val, y_val))
    print("-" * 35 + "\n")
```
